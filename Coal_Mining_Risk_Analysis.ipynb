```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)


# =========================================================
# 1. LOAD DATASET
# =========================================================

df = pd.read_csv("coal_mining_risk_dataset_final.csv")

print("Dataset shape:", df.shape)


# =========================================================
# 2. FEATURES AND TARGET
# =========================================================

# Target
y = df["Risk_Level"]


# Remove target and columns that should NOT be used
X = df.drop(
    columns=[
        "Risk_Level",
        "Risk_Score",
        "Year"
    ]
)


print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())


# =========================================================
# 3. IDENTIFY CATEGORICAL / NUMERICAL FEATURES
# =========================================================

categorical_features = [
    "State",
    "District",
    "Mine_Name",
    "Mine_Type",
    "Place_of_Operation",
    "Accident_Type",
    "Machinery_Involved",
    "Worker_Experience_Years",
    "Training_Level",
    "Shift"
]

numerical_features = [
    "Injured",
    "Death",
    "Worker_Age",
    "Safety_Compliance_Score"
]


# =========================================================
# 4. PREPROCESSING
# =========================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)


# =========================================================
# 5. TRAIN / TEST SPLIT
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# =========================================================
# 6. MODELS
# =========================================================

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            max_depth=8,
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            max_depth=10,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            n_estimators=150,
            learning_rate=0.08,
            max_depth=3,
            random_state=42
        )
}


# =========================================================
# 7. TRAIN AND EVALUATE
# =========================================================

results = []

trained_models = {}

for name, model in models.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    # Train
    pipeline.fit(X_train, y_train)

    # Predict
    predictions = pipeline.predict(X_test)

    # Metrics
    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted",
        zero_division=0
    )

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            predictions,
            zero_division=0
        )
    )

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])

    trained_models[name] = pipeline


# =========================================================
# 8. MODEL COMPARISON
# =========================================================

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ]
)

print("\n\n========== MODEL COMPARISON ==========")
print(
    results_df
    .sort_values(
        "F1_Score",
        ascending=False
    )
    .to_string(index=False)
)


# =========================================================
# 9. SELECT BEST MODEL
# =========================================================

best_model_name = (
    results_df
    .sort_values(
        "F1_Score",
        ascending=False
    )
    .iloc[0]["Model"]
)

best_model = trained_models[best_model_name]

print("\nBest Model:", best_model_name)


# =========================================================
# 10. SAVE BEST MODEL
# =========================================================

import joblib

joblib.dump(
    best_model,
    "coal_mining_risk_model.pkl"
)

print("\nModel saved as:")
print("coal_mining_risk_model.pkl")
```

```python
# =========================================================
# CELL 12 - FIX NONE VALUES + CREATE PRE-ACCIDENT RISK
# =========================================================

# These columns legitimately contain the category "None"
# Replace NaN values caused by pandas interpreting "None" as missing
df["Training_Level"] = df["Training_Level"].fillna("None")
df["Machinery_Involved"] = df["Machinery_Involved"].fillna("None")


# Check that the problem is fixed
print("Training Level:")
print(df["Training_Level"].value_counts())

print("\nMachinery:")
print(df["Machinery_Involved"].value_counts())


# =========================================================
# CREATE RISK SCORE
# =========================================================

df["Pre_Accident_Risk_Score"] = 0.0


# ---------------------------------------------------------
# 1. SAFETY COMPLIANCE
# ---------------------------------------------------------

df["Pre_Accident_Risk_Score"] += (
    (100 - df["Safety_Compliance_Score"]) * 0.50
)


# ---------------------------------------------------------
# 2. MINE TYPE
# ---------------------------------------------------------

df["Pre_Accident_Risk_Score"] += np.where(
    df["Mine_Type"] == "Underground",
    15,
    0
)


# ---------------------------------------------------------
# 3. MACHINERY
# ---------------------------------------------------------

machinery_risk = {
    "None": 0,
    "Drill": 4,
    "Excavator": 7,
    "Loader": 10,
    "Dumper": 13,
    "Truck": 16,
    "Heavy Earth Moving Machinery": 19
}

df["Pre_Accident_Risk_Score"] += (
    df["Machinery_Involved"].map(machinery_risk)
)


# ---------------------------------------------------------
# 4. TRAINING
# ---------------------------------------------------------

training_risk = {
    "None": 15,
    "Basic": 11,
    "Intermediate": 7,
    "Advanced": 3,
    "Certified": 0
}

df["Pre_Accident_Risk_Score"] += (
    df["Training_Level"].map(training_risk)
)


# ---------------------------------------------------------
# 5. EXPERIENCE
# ---------------------------------------------------------

experience_risk = {
    "<1 year": 12,
    "1–3 years": 9,
    "3–5 years": 6,
    "5–10 years": 3,
    "10+ years": 0
}

df["Pre_Accident_Risk_Score"] += (
    df["Worker_Experience_Years"].map(experience_risk)
)


# ---------------------------------------------------------
# 6. SHIFT
# ---------------------------------------------------------

shift_risk = {
    "Morning": 0,
    "Afternoon": 5,
    "Night": 10
}

df["Pre_Accident_Risk_Score"] += (
    df["Shift"].map(shift_risk)
)


# ---------------------------------------------------------
# 7. PLACE OF OPERATION
# ---------------------------------------------------------

place_risk = {
    "Face": 10,
    "Haul Road": 13,
    "Workshop": 6,
    "Stockyard": 7,
    "Shaft": 11,
    "Excavation Area": 9
}

df["Pre_Accident_Risk_Score"] += (
    df["Place_of_Operation"].map(place_risk)
)


# ---------------------------------------------------------
# 8. WORKER AGE
# ---------------------------------------------------------

df["Pre_Accident_Risk_Score"] += np.where(
    df["Worker_Age"] < 21,
    4,
    np.where(
        df["Worker_Age"] > 50,
        3,
        0
    )
)


# ---------------------------------------------------------
# 9. CONTROLLED RANDOMNESS
# ---------------------------------------------------------

np.random.seed(42)

df["Pre_Accident_Risk_Score"] += np.random.normal(
    0,
    2,
    len(df)
)


print("\nPre-Accident Risk Score created successfully.")

print("\nRisk Score Statistics:")

print(
    df["Pre_Accident_Risk_Score"].describe()
)
```

```python
# =========================================================
# CELL 13 - CREATE PRE-ACCIDENT RISK LEVEL
# =========================================================

# Calculate quartile thresholds
q1 = df["Pre_Accident_Risk_Score"].quantile(0.25)
q2 = df["Pre_Accident_Risk_Score"].quantile(0.50)
q3 = df["Pre_Accident_Risk_Score"].quantile(0.75)


def classify_pre_accident_risk(score):

    if score <= q1:
        return "Low"

    elif score <= q2:
        return "Moderate"

    elif score <= q3:
        return "High"

    else:
        return "Very High"


df["Pre_Accident_Risk_Level"] = (
    df["Pre_Accident_Risk_Score"]
    .apply(classify_pre_accident_risk)
)


print("========== PRE-ACCIDENT RISK DISTRIBUTION ==========\n")

print(
    df["Pre_Accident_Risk_Level"].value_counts()
)


print("\n========== PERCENTAGE ==========\n")

print(
    df["Pre_Accident_Risk_Level"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)


print("\n========== RISK THRESHOLDS ==========")

print("Low threshold      :", round(q1, 2))
print("Moderate threshold :", round(q2, 2))
print("High threshold     :", round(q3, 2))
```

```python
# =========================================================
# CELL 14 - VALIDATE PRE-ACCIDENT RISK RELATIONSHIPS
# =========================================================

print("========== RISK BY MINE TYPE ==========\n")

print(
    df.groupby("Mine_Type")[
        "Pre_Accident_Risk_Score"
    ].mean()
)


print("\n========== RISK BY SHIFT ==========\n")

print(
    df.groupby("Shift")[
        "Pre_Accident_Risk_Score"
    ].mean()
    .reindex([
        "Morning",
        "Afternoon",
        "Night"
    ])
)


print("\n========== RISK BY TRAINING ==========\n")

training_order = [
    "None",
    "Basic",
    "Intermediate",
    "Advanced",
    "Certified"
]

print(
    df.groupby("Training_Level")[
        "Pre_Accident_Risk_Score"
    ].mean()
    .reindex(training_order)
)


print("\n========== RISK BY EXPERIENCE ==========\n")

experience_order = [
    "<1 year",
    "1–3 years",
    "3–5 years",
    "5–10 years",
    "10+ years"
]

print(
    df.groupby("Worker_Experience_Years")[
        "Pre_Accident_Risk_Score"
    ].mean()
    .reindex(experience_order)
)


print("\n========== RISK BY MACHINERY ==========\n")

machinery_order = [
    "None",
    "Drill",
    "Excavator",
    "Loader",
    "Dumper",
    "Truck",
    "Heavy Earth Moving Machinery"
]

print(
    df.groupby("Machinery_Involved")[
        "Pre_Accident_Risk_Score"
    ].mean()
    .reindex(machinery_order)
)
```

```python
# =========================================================
# CELL 16 - FINAL PRE-ACCIDENT ML MODEL
# =========================================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

import pandas as pd


# =========================================================
# 1. FEATURES AND TARGET
# =========================================================

prediction_features = [
    "State",
    "District",
    "Mine_Name",
    "Mine_Type",
    "Place_of_Operation",
    "Machinery_Involved",
    "Worker_Age",
    "Worker_Experience_Years",
    "Training_Level",
    "Shift",
    "Safety_Compliance_Score"
]

X = df[prediction_features]

y = df["Pre_Accident_Risk_Level"]


print("========== FINAL MODEL FEATURES ==========\n")

print(prediction_features)

print("\n========== TARGET DISTRIBUTION ==========\n")

print(y.value_counts())


# =========================================================
# 2. CATEGORICAL FEATURES
# =========================================================

categorical_features = [
    "State",
    "District",
    "Mine_Name",
    "Mine_Type",
    "Place_of_Operation",
    "Machinery_Involved",
    "Worker_Experience_Years",
    "Training_Level",
    "Shift"
]


# =========================================================
# 3. NUMERICAL FEATURES
# =========================================================

numerical_features = [
    "Worker_Age",
    "Safety_Compliance_Score"
]


# =========================================================
# 4. PREPROCESSING
# =========================================================

preprocessor_final = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)


# =========================================================
# 5. TRAIN / TEST SPLIT
# =========================================================

X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\nTraining samples:", len(X_train_final))
print("Testing samples:", len(X_test_final))


# =========================================================
# 6. MODELS
# =========================================================

final_models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=2000
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            max_depth=8,
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=200,
            max_depth=10,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            n_estimators=150,
            learning_rate=0.08,
            max_depth=3,
            random_state=42
        )
}


# =========================================================
# 7. TRAIN MODELS
# =========================================================

final_results = []

final_trained_models = {}


for name, model in final_models.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor_final
            ),
            (
                "model",
                model
            )
        ]
    )


    # Train
    pipeline.fit(
        X_train_final,
        y_train_final
    )


    # Predict
    predictions = pipeline.predict(
        X_test_final
    )


    # Metrics
    accuracy = accuracy_score(
        y_test_final,
        predictions
    )

    precision = precision_score(
        y_test_final,
        predictions,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test_final,
        predictions,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test_final,
        predictions,
        average="weighted",
        zero_division=0
    )


    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))


    final_results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])


    final_trained_models[name] = pipeline


# =========================================================
# 8. MODEL COMPARISON
# =========================================================

final_results_df = pd.DataFrame(
    final_results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ]
)


print("\n\n========== FINAL MODEL COMPARISON ==========\n")


print(
    final_results_df
    .sort_values(
        "F1_Score",
        ascending=False
    )
    .to_string(index=False)
)


# =========================================================
# 9. SELECT BEST MODEL
# =========================================================

best_final_model_name = (
    final_results_df
    .sort_values(
        "F1_Score",
        ascending=False
    )
    .iloc[0]["Model"]
)


best_final_model = (
    final_trained_models[
        best_final_model_name
    ]
)


print(
    "\nBest Final Model:",
    best_final_model_name
)
```

```python
# =========================================================
# CELL 17 - FINAL MODEL CONFUSION MATRIX
# =========================================================

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Predict using final model
y_pred_final = best_final_model.predict(X_test_final)

# Risk class order
risk_labels = [
    "Low",
    "Moderate",
    "High",
    "Very High"
]

# Create confusion matrix
cm_final = confusion_matrix(
    y_test_final,
    y_pred_final,
    labels=risk_labels
)

# Plot
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm_final,
    annot=True,
    fmt="d",
    xticklabels=risk_labels,
    yticklabels=risk_labels
)

plt.title("Final Pre-Accident Risk Prediction")
plt.xlabel("Predicted Risk Level")
plt.ylabel("Actual Risk Level")

plt.tight_layout()
plt.show()
```

```python
# =========================================================
# CELL 18 - FINAL CLASSIFICATION REPORT
# =========================================================

print("========== FINAL MODEL PERFORMANCE ==========\n")

print(
    classification_report(
        y_test_final,
        y_pred_final,
        labels=risk_labels,
        zero_division=0
    )
)
```

```python
# =========================================================
# CELL 19 - FINAL FEATURE IMPORTANCE
# =========================================================

import numpy as np

# Get preprocessing and ML model
final_preprocessor = best_final_model.named_steps["preprocessor"]
final_ml_model = best_final_model.named_steps["model"]

# Get encoded feature names
final_feature_names = (
    final_preprocessor
    .get_feature_names_out()
)


# =========================================================
# GET IMPORTANCE
# =========================================================

if hasattr(final_ml_model, "feature_importances_"):

    # Tree-based models
    final_importances = (
        final_ml_model
        .feature_importances_
    )

elif hasattr(final_ml_model, "coef_"):

    # Logistic Regression
    final_importances = np.mean(
        np.abs(
            final_ml_model.coef_
        ),
        axis=0
    )

else:

    raise ValueError(
        "This model does not support feature importance."
    )


# =========================================================
# CREATE DATAFRAME
# =========================================================

final_feature_importance = pd.DataFrame({

    "Feature": final_feature_names,

    "Importance": final_importances

})


final_feature_importance = (
    final_feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
)


print("========== TOP 20 FINAL FEATURES ==========\n")

print(
    final_feature_importance
    .head(20)
    .to_string(index=False)
)
```

```python
%%writefile app.py

from flask import Flask, request, jsonify, render_template
from flask_cors import CORS
import pandas as pd
import joblib
import os


# =========================================================
# FLASK APPLICATION
# =========================================================

app = Flask(__name__)
CORS(app)


# =========================================================
# LOAD ML MODEL
# =========================================================

MODEL_FILE = "final_coal_mining_risk_model.pkl"

model = joblib.load(MODEL_FILE)


# =========================================================
# LOAD DATASET
# Used for dropdown options
# =========================================================

DATASET_FILE = "coal_mining_risk_ml_dataset.csv"

df = pd.read_csv(DATASET_FILE)


# Fix None categories
df["Training_Level"] = (
    df["Training_Level"]
    .fillna("None")
)

df["Machinery_Involved"] = (
    df["Machinery_Involved"]
    .fillna("None")
)


# =========================================================
# HOME PAGE
# =========================================================

@app.route("/")
def home():

    return render_template(
        "index.html"
    )


# =========================================================
# OPTIONS API
# =========================================================

@app.route("/options", methods=["GET"])
def options():

    return jsonify({

        "states": sorted(
            df["State"]
            .dropna()
            .unique()
            .tolist()
        ),

        "districts": sorted(
            df["District"]
            .dropna()
            .unique()
            .tolist()
        ),

        "mines": sorted(
            df["Mine_Name"]
            .dropna()
            .unique()
            .tolist()
        ),

        "mine_types": [
            "Underground",
            "Opencast"
        ],

        "places": [
            "Face",
            "Haul Road",
            "Workshop",
            "Stockyard",
            "Shaft",
            "Excavation Area"
        ],

        "machinery": [
            "None",
            "Drill",
            "Excavator",
            "Loader",
            "Dumper",
            "Truck",
            "Heavy Earth Moving Machinery"
        ],

        "experience": [
            "<1 year",
            "1–3 years",
            "3–5 years",
            "5–10 years",
            "10+ years"
        ],

        "training": [
            "None",
            "Basic",
            "Intermediate",
            "Advanced",
            "Certified"
        ],

        "shifts": [
            "Morning",
            "Afternoon",
            "Night"
        ]
    })


# =========================================================
# PREDICTION API
# =========================================================

@app.route("/predict", methods=["POST"])
def predict():

    try:

        data = request.get_json()


        # -------------------------------------------------
        # Validate required fields
        # -------------------------------------------------

        required_fields = [
            "State",
            "District",
            "Mine_Name",
            "Mine_Type",
            "Place_of_Operation",
            "Machinery_Involved",
            "Worker_Age",
            "Worker_Experience_Years",
            "Training_Level",
            "Shift",
            "Safety_Compliance_Score"
        ]


        for field in required_fields:

            if field not in data:

                return jsonify({
                    "error": f"Missing field: {field}"
                }), 400


        # -------------------------------------------------
        # Create input dataframe
        # -------------------------------------------------

        input_data = pd.DataFrame({

            "State": [
                data["State"]
            ],

            "District": [
                data["District"]
            ],

            "Mine_Name": [
                data["Mine_Name"]
            ],

            "Mine_Type": [
                data["Mine_Type"]
            ],

            "Place_of_Operation": [
                data["Place_of_Operation"]
            ],

            "Machinery_Involved": [
                data["Machinery_Involved"]
            ],

            "Worker_Age": [
                float(data["Worker_Age"])
            ],

            "Worker_Experience_Years": [
                data["Worker_Experience_Years"]
            ],

            "Training_Level": [
                data["Training_Level"]
            ],

            "Shift": [
                data["Shift"]
            ],

            "Safety_Compliance_Score": [
                float(data["Safety_Compliance_Score"])
            ]
        })


        # -------------------------------------------------
        # Prediction
        # -------------------------------------------------

        prediction = model.predict(
            input_data
        )[0]


        # -------------------------------------------------
        # Probability
        # -------------------------------------------------

        probabilities = model.predict_proba(
            input_data
        )[0]


        classes = model.classes_


        probability_dict = {

            str(cls): round(
                float(prob) * 100,
                2
            )

            for cls, prob
            in zip(classes, probabilities)
        }


        # -------------------------------------------------
        # Highest confidence
        # -------------------------------------------------

        confidence = round(
            float(max(probabilities)) * 100,
            2
        )


        # -------------------------------------------------
        # Response
        # -------------------------------------------------

        return jsonify({

            "risk_level": prediction,

            "confidence": confidence,

            "probabilities": probability_dict

        })


    except Exception as e:

        return jsonify({

            "error": str(e)

        }), 500


# =========================================================
# RUN SERVER
# =========================================================

if __name__ == "__main__":

    app.run(
        host="127.0.0.1",
        port=5000,
        debug=True
    )
```

```python
%%writefile templates/index.html

<!DOCTYPE html>

<html lang="en">

<head>

    <meta charset="UTF-8">

    <meta
        name="viewport"
        content="width=device-width, initial-scale=1.0"
    >

    <title>
        Coal Mining Risk Analysis
    </title>


    <style>

        * {
            box-sizing: border-box;
        }


        body {

            margin: 0;

            font-family:
                Arial,
                Helvetica,
                sans-serif;

            background:
                linear-gradient(
                    135deg,
                    #eef2f3,
                    #dfe9f3
                );

            color: #1f2937;
        }


        .container {

            width: 92%;

            max-width: 1100px;

            margin: 40px auto;

            background: white;

            padding: 35px;

            border-radius: 18px;

            box-shadow:
                0 10px 35px
                rgba(0, 0, 0, 0.12);
        }


        h1 {

            margin-top: 0;

            font-size: 32px;

            color: #243447;
        }


        .subtitle {

            color: #6b7280;

            margin-bottom: 30px;
        }


        .section-title {

            font-size: 24px;

            margin-top: 10px;

            margin-bottom: 25px;
        }


        .form-grid {

            display: grid;

            grid-template-columns:
                repeat(3, 1fr);

            gap: 22px;
        }


        .field {

            display: flex;

            flex-direction: column;
        }


        label {

            font-size: 14px;

            margin-bottom: 8px;

            color: #374151;

            font-weight: 600;
        }


        input,
        select {

            width: 100%;

            padding: 13px;

            border: 1px solid #d1d5db;

            border-radius: 9px;

            background: #f8fafc;

            font-size: 15px;

            outline: none;

            transition: 0.2s;
        }


        input:focus,
        select:focus {

            border-color: #2563eb;

            background: white;
        }


        .button-container {

            margin-top: 30px;

            display: flex;

            justify-content: center;
        }


        button {

            padding:
                14px 28px;

            border: none;

            border-radius: 9px;

            background: #ef4444;

            color: white;

            font-size: 16px;

            font-weight: 600;

            cursor: pointer;

            transition: 0.2s;
        }


        button:hover {

            background: #dc2626;

            transform:
                translateY(-1px);
        }


        button:disabled {

            opacity: 0.6;

            cursor: not-allowed;

            transform: none;
        }


        .result {

            margin-top: 40px;

            padding-top: 25px;

            border-top:
                1px solid #e5e7eb;

            display: none;
        }


        .result-title {

            font-size: 25px;

            margin-bottom: 25px;
        }


        .result-grid {

            display: grid;

            grid-template-columns:
                1fr 1fr;

            gap: 25px;
        }


        .result-card {

            padding: 20px;

            border-radius: 12px;

            background: #f8fafc;
        }


        .result-label {

            font-size: 14px;

            color: #6b7280;

            margin-bottom: 8px;
        }


        .risk-value {

            font-size: 34px;

            font-weight: 700;
        }


        .confidence-value {

            font-size: 30px;

            font-weight: 700;
        }


        .message {

            margin-top: 20px;

            padding: 16px;

            border-radius: 10px;

            background: #ecfdf5;

            color: #15803d;

            font-weight: 500;
        }


        .probabilities {

            margin-top: 25px;
        }


        .probability-row {

            display: flex;

            justify-content:
                space-between;

            padding: 9px 0;

            border-bottom:
                1px solid #e5e7eb;
        }


        .error {

            margin-top: 20px;

            padding: 15px;

            border-radius: 9px;

            background: #fef2f2;

            color: #b91c1c;

            display: none;
        }


        @media (max-width: 800px) {

            .form-grid {

                grid-template-columns:
                    1fr;
            }


            .result-grid {

                grid-template-columns:
                    1fr;
            }

            .container {

                padding: 22px;
            }
        }

    </style>

</head>


<body>


<div class="container">


    <h1>
        Coal Mining Risk Analysis
    </h1>


    <p class="subtitle">
        Machine Learning Based Accident Risk Prediction System
    </p>


    <h2 class="section-title">
        Mining Conditions
    </h2>


    <form id="predictionForm">


        <div class="form-grid">


            <!-- STATE -->

            <div class="field">

                <label>
                    State
                </label>

                <select
                    id="State"
                    required
                >

                    <option value="">
                        Select State
                    </option>

                </select>

            </div>


            <!-- SHIFT -->

            <div class="field">

                <label>
                    Shift
                </label>

                <select
                    id="Shift"
                    required
                >

                    <option value="">
                        Select Shift
                    </option>

                </select>

            </div>


            <!-- EXPERIENCE -->

            <div class="field">

                <label>
                    Experience (Years)
                </label>

                <select
                    id="Worker_Experience_Years"
                    required
                >

                    <option value="">
                        Select Experience
                    </option>

                </select>

            </div>


            <!-- DISTRICT -->

            <div class="field">

                <label>
                    District
                </label>

                <select
                    id="District"
                    required
                >

                    <option value="">
                        Select District
                    </option>

                </select>

            </div>


            <!-- MINE TYPE -->

            <div class="field">

                <label>
                    Mine Type
                </label>

                <select
                    id="Mine_Type"
                    required
                >

                    <option value="">
                        Select Mine Type
                    </option>

                </select>

            </div>


            <!-- MACHINERY -->

            <div class="field">

                <label>
                    Machinery Involved
                </label>

                <select
                    id="Machinery_Involved"
                    required
                >

                    <option value="">
                        Select Machinery
                    </option>

                </select>

            </div>


            <!-- MINE NAME -->

            <div class="field">

                <label>
                    Mine Name
                </label>

                <select
                    id="Mine_Name"
                    required
                >

                    <option value="">
                        Select Mine
                    </option>

                </select>

            </div>


            <!-- PLACE -->

            <div class="field">

                <label>
                    Place of Operation
                </label>

                <select
                    id="Place_of_Operation"
                    required
                >

                    <option value="">
                        Select Place
                    </option>

                </select>

            </div>


            <!-- SAFETY -->

            <div class="field">

                <label>
                    Safety Compliance Score
                </label>

                <input
                    type="number"
                    id="Safety_Compliance_Score"
                    min="0"
                    max="100"
                    step="0.01"
                    placeholder="0 - 100"
                    required
                >

            </div>


            <!-- TRAINING -->

            <div class="field">

                <label>
                    Training Level
                </label>

                <select
                    id="Training_Level"
                    required
                >

                    <option value="">
                        Select Training
                    </option>

                </select>

            </div>


            <!-- AGE -->

            <div class="field">

                <label>
                    Worker Age
                </label>

                <input
                    type="number"
                    id="Worker_Age"
                    min="18"
                    max="70"
                    placeholder="Worker age"
                    required
                >

            </div>


        </div>


        <div class="button-container">

            <button
                type="submit"
                id="predictButton"
            >

                Predict Accident Risk

            </button>

        </div>


    </form>


    <div
        class="error"
        id="errorMessage"
    ></div>


    <!-- RESULT -->

    <div
        class="result"
        id="resultSection"
    >


        <h2 class="result-title">
            Prediction Result
        </h2>


        <div class="result-grid">


            <div class="result-card">

                <div class="result-label">
                    Predicted Risk
                </div>

                <div
                    class="risk-value"
                    id="riskLevel"
                >
                    --
                </div>

            </div>


            <div class="result-card">

                <div class="result-label">
                    Prediction Confidence
                </div>

                <div
                    class="confidence-value"
                    id="confidence"
                >
                    --
                </div>

            </div>


        </div>


        <div
            class="message"
            id="resultMessage"
        ></div>


        <div class="probabilities">

            <h3>
                Risk Probabilities
            </h3>


            <div
                class="probability-row"
            >

                <span>
                    Low
                </span>

                <strong id="lowProbability">
                    --
                </strong>

            </div>


            <div
                class="probability-row"
            >

                <span>
                    Moderate
                </span>

                <strong id="moderateProbability">
                    --
                </strong>

            </div>


            <div
                class="probability-row"
            >

                <span>
                    High
                </span>

                <strong id="highProbability">
                    --
                </strong>

            </div>


            <div
                class="probability-row"
            >

                <span>
                    Very High
                </span>

                <strong id="veryHighProbability">
                    --
                </strong>

            </div>


        </div>


    </div>


</div>


<script>


// =========================================================
// LOAD OPTIONS
// =========================================================

async function loadOptions() {

    try {

        const response =
            await fetch("/options");

        const data =
            await response.json();


        fillSelect(
            "State",
            data.states
        );

        fillSelect(
            "District",
            data.districts
        );

        fillSelect(
            "Mine_Name",
            data.mines
        );

        fillSelect(
            "Mine_Type",
            data.mine_types
        );

        fillSelect(
            "Place_of_Operation",
            data.places
        );

        fillSelect(
            "Machinery_Involved",
            data.machinery
        );

        fillSelect(
            "Worker_Experience_Years",
            data.experience
        );

        fillSelect(
            "Training_Level",
            data.training
        );

        fillSelect(
            "Shift",
            data.shifts
        );

    }

    catch (error) {

        showError(
            "Unable to load mining options."
        );

    }

}


// =========================================================
// FILL SELECT
// =========================================================

function fillSelect(
    elementId,
    values
) {

    const select =
        document.getElementById(
            elementId
        );


    values.forEach(
        value => {

            const option =
                document.createElement(
                    "option"
                );

            option.value = value;

            option.textContent = value;

            select.appendChild(
                option
            );

        }
    );

}


// =========================================================
// FORM SUBMISSION
// =========================================================

document
    .getElementById(
        "predictionForm"
    )
    .addEventListener(
        "submit",
        async function(event) {

            event.preventDefault();


            const button =
                document.getElementById(
                    "predictButton"
                );


            button.disabled = true;

            button.textContent =
                "Predicting...";


            hideError();


            const data = {

                State:
                    document.getElementById(
                        "State"
                    ).value,

                District:
                    document.getElementById(
                        "District"
                    ).value,

                Mine_Name:
                    document.getElementById(
                        "Mine_Name"
                    ).value,

                Mine_Type:
                    document.getElementById(
                        "Mine_Type"
                    ).value,

                Place_of_Operation:
                    document.getElementById(
                        "Place_of_Operation"
                    ).value,

                Machinery_Involved:
                    document.getElementById(
                        "Machinery_Involved"
                    ).value,

                Worker_Age:
                    document.getElementById(
                        "Worker_Age"
                    ).value,

                Worker_Experience_Years:
                    document.getElementById(
                        "Worker_Experience_Years"
                    ).value,

                Training_Level:
                    document.getElementById(
                        "Training_Level"
                    ).value,

                Shift:
                    document.getElementById(
                        "Shift"
                    ).value,

                Safety_Compliance_Score:
                    document.getElementById(
                        "Safety_Compliance_Score"
                    ).value
            };


            try {

                const response =
                    await fetch(
                        "/predict",
                        {

                            method: "POST",

                            headers: {
                                "Content-Type":
                                    "application/json"
                            },

                            body:
                                JSON.stringify(data)

                        }
                    );


                const result =
                    await response.json();


                if (!response.ok) {

                    throw new Error(
                        result.error ||
                        "Prediction failed."
                    );

                }


                // -----------------------------------------
                // DISPLAY RESULT
                // -----------------------------------------

                const risk =
                    result.risk_level;


                document
                    .getElementById(
                        "riskLevel"
                    )
                    .textContent =
                    risk.toUpperCase() +
                    " RISK";


                document
                    .getElementById(
                        "confidence"
                    )
                    .textContent =
                    result.confidence +
                    "%";


                document
                    .getElementById(
                        "resultMessage"
                    )
                    .textContent =
                    "Prediction completed: " +
                    risk.toUpperCase() +
                    " RISK";


                // -----------------------------------------
                // PROBABILITIES
                // -----------------------------------------

                document
                    .getElementById(
                        "lowProbability"
                    )
                    .textContent =
                    (
                        result.probabilities.Low ||
                        0
                    ) + "%";


                document
                    .getElementById(
                        "moderateProbability"
                    )
                    .textContent =
                    (
                        result.probabilities.Moderate ||
                        0
                    ) + "%";


                document
                    .getElementById(
                        "highProbability"
                    )
                    .textContent =
                    (
                        result.probabilities.High ||
                        0
                    ) + "%";


                document
                    .getElementById(
                        "veryHighProbability"
                    )
                    .textContent =
                    (
                        result.probabilities[
                            "Very High"
                        ] ||
                        0
                    ) + "%";


                document
                    .getElementById(
                        "resultSection"
                    )
                    .style.display =
                    "block";


                document
                    .getElementById(
                        "resultSection"
                    )
                    .scrollIntoView({
                        behavior: "smooth"
                    });


            }

            catch (error) {

                showError(
                    error.message
                );

            }


            button.disabled = false;

            button.textContent =
                "Predict Accident Risk";

        }
    );


// =========================================================
// ERROR HANDLING
// =========================================================

function showError(message) {

    const error =
        document.getElementById(
            "errorMessage"
        );

    error.textContent = message;

    error.style.display = "block";

}


function hideError() {

    document
        .getElementById(
            "errorMessage"
        )
        .style.display = "none";

}


// =========================================================
// INITIALIZE
// =========================================================

loadOptions();

</script>


</body>

</html>
```

```python
# =========================================================
# CELL 34 - CREATE DASHBOARD
# =========================================================

dashboard_html = r'''
<!DOCTYPE html>

<html lang="en">

<head>

<meta charset="UTF-8">

<meta
    name="viewport"
    content="width=device-width, initial-scale=1.0"
>

<title>Mining Risk Dashboard</title>

<script src="https://cdn.jsdelivr.net/npm/chart.js"></script>

<style>

* {
    box-sizing: border-box;
}

body {

    margin: 0;

    font-family:
        Arial,
        Helvetica,
        sans-serif;

    background: #eef3f7;

    color: #172033;
}

.container {

    width: 94%;

    max-width: 1250px;

    margin: 30px auto;
}

.header {

    background: white;

    padding: 28px 32px;

    border-radius: 16px;

    box-shadow:
        0 5px 20px
        rgba(0,0,0,0.08);

    margin-bottom: 22px;
}

.header h1 {

    margin: 0 0 8px;

    font-size: 30px;
}

.header p {

    margin: 0;

    color: #64748b;
}

.nav {

    margin-top: 18px;
}

.nav a {

    display: inline-block;

    padding: 10px 18px;

    background: #ef4444;

    color: white;

    text-decoration: none;

    border-radius: 8px;

    font-weight: 600;
}

.stats {

    display: grid;

    grid-template-columns:
        repeat(4, 1fr);

    gap: 18px;

    margin-bottom: 22px;
}

.stat {

    background: white;

    padding: 22px;

    border-radius: 14px;

    box-shadow:
        0 4px 15px
        rgba(0,0,0,0.06);
}

.stat-title {

    color: #64748b;

    font-size: 14px;

    margin-bottom: 10px;
}

.stat-value {

    font-size: 30px;

    font-weight: 700;
}

.grid {

    display: grid;

    grid-template-columns:
        repeat(2, 1fr);

    gap: 22px;
}

.card {

    background: white;

    padding: 22px;

    border-radius: 14px;

    box-shadow:
        0 4px 15px
        rgba(0,0,0,0.06);
}

.card h2 {

    margin-top: 0;

    font-size: 20px;
}

.chart-container {

    height: 320px;

    position: relative;
}

.full {

    grid-column: 1 / -1;
}

@media(max-width: 800px) {

    .stats {

        grid-template-columns: 1fr 1fr;
    }

    .grid {

        grid-template-columns: 1fr;
    }

}

@media(max-width: 500px) {

    .stats {

        grid-template-columns: 1fr;
    }

}

</style>

</head>


<body>


<div class="container">


<div class="header">

    <h1>
        Coal Mining Risk Analysis Dashboard
    </h1>

    <p>
        Historical accident analysis and pre-accident risk overview
    </p>

    <div class="nav">

        <a href="/">
            ← Risk Prediction
        </a>

    </div>

</div>


<!-- =====================================================
     STATISTICS
====================================================== -->

<div class="stats">


    <div class="stat">

        <div class="stat-title">
            Total Records
        </div>

        <div
            class="stat-value"
            id="totalRecords"
        >
            --
        </div>

    </div>


    <div class="stat">

        <div class="stat-title">
            Total Injured
        </div>

        <div
            class="stat-value"
            id="totalInjured"
        >
            --
        </div>

    </div>


    <div class="stat">

        <div class="stat-title">
            Total Deaths
        </div>

        <div
            class="stat-value"
            id="totalDeaths"
        >
            --
        </div>

    </div>


    <div class="stat">

        <div class="stat-title">
            Average Safety Score
        </div>

        <div
            class="stat-value"
            id="averageSafety"
        >
            --
        </div>

    </div>


</div>


<!-- =====================================================
     CHARTS
====================================================== -->

<div class="grid">


    <div class="card full">

        <h2>
            Accidents by State
        </h2>

        <div class="chart-container">

            <canvas id="stateChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Accidents by Year
        </h2>

        <div class="chart-container">

            <canvas id="yearChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Risk Level Distribution
        </h2>

        <div class="chart-container">

            <canvas id="riskChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Mine Type
        </h2>

        <div class="chart-container">

            <canvas id="mineTypeChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Shift Analysis
        </h2>

        <div class="chart-container">

            <canvas id="shiftChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Training Level
        </h2>

        <div class="chart-container">

            <canvas id="trainingChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Machinery Involved
        </h2>

        <div class="chart-container">

            <canvas id="machineryChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Place of Operation
        </h2>

        <div class="chart-container">

            <canvas id="placeChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Accident Types
        </h2>

        <div class="chart-container">

            <canvas id="accidentChart"></canvas>

        </div>

    </div>


    <div class="card">

        <h2>
            Average Safety Score by Risk
        </h2>

        <div class="chart-container">

            <canvas id="safetyChart"></canvas>

        </div>

    </div>


</div>


</div>


<script>


// =========================================================
// LOAD DASHBOARD
// =========================================================

async function loadDashboard() {

    const response =
        await fetch("/dashboard-data");

    const data =
        await response.json();


    // Statistics

    document
        .getElementById("totalRecords")
        .textContent =
        data.total_records;

    document
        .getElementById("totalInjured")
        .textContent =
        data.total_injured;

    document
        .getElementById("totalDeaths")
        .textContent =
        data.total_deaths;

    document
        .getElementById("averageSafety")
        .textContent =
        data.average_safety;


    // Charts

    createBarChart(
        "stateChart",
        data.state_counts,
        "Accidents"
    );


    createLineChart(
        "yearChart",
        data.year_counts
    );


    createDoughnutChart(
        "riskChart",
        data.risk_counts
    );


    createDoughnutChart(
        "mineTypeChart",
        data.mine_type_counts
    );


    createBarChart(
        "shiftChart",
        data.shift_counts,
        "Accidents"
    );


    createBarChart(
        "trainingChart",
        data.training_counts,
        "Accidents"
    );


    createBarChart(
        "machineryChart",
        data.machinery_counts,
        "Accidents"
    );


    createBarChart(
        "placeChart",
        data.place_counts,
        "Accidents"
    );


    createBarChart(
        "accidentChart",
        data.accident_counts,
        "Accidents"
    );


    createBarChart(
        "safetyChart",
        data.safety_by_risk,
        "Average Safety Score"
    );

}


// =========================================================
// BAR CHART
// =========================================================

function createBarChart(
    canvasId,
    object,
    label
) {

    const labels =
        Object.keys(object);

    const values =
        Object.values(object);


    new Chart(
        document.getElementById(canvasId),
        {

            type: "bar",

            data: {

                labels: labels,

                datasets: [

                    {
                        label: label,

                        data: values
                    }

                ]
            },

            options: {

                responsive: true,

                maintainAspectRatio: false,

                plugins: {

                    legend: {
                        display: false
                    }

                }
            }

        }
    );

}


// =========================================================
// LINE CHART
// =========================================================

function createLineChart(
    canvasId,
    object
) {

    new Chart(

        document.getElementById(canvasId),

        {

            type: "line",

            data: {

                labels:
                    Object.keys(object),

                datasets: [

                    {

                        label:
                            "Accidents",

                        data:
                            Object.values(object),

                        tension: 0.3,

                        fill: false

                    }

                ]

            },

            options: {

                responsive: true,

                maintainAspectRatio: false

            }

        }

    );

}


// =========================================================
// DOUGHNUT CHART
// =========================================================

function createDoughnutChart(
    canvasId,
    object
) {

    new Chart(

        document.getElementById(canvasId),

        {

            type: "doughnut",

            data: {

                labels:
                    Object.keys(object),

                datasets: [

                    {

                        data:
                            Object.values(object)

                    }

                ]

            },

            options: {

                responsive: true,

                maintainAspectRatio: false

            }

        }

    );

}


loadDashboard();

</script>


</body>

</html>
'''

Path(
    "templates/dashboard.html"
).write_text(
    dashboard_html,
    encoding="utf-8"
)

print("Dashboard created successfully!")
```
